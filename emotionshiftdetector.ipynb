```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
from transformers import pipeline

emotion_model = pipeline(
    "text-classification",
    model="SamLowe/roberta-base-go_emotions",
    top_k=1
)
#uses roberta which handles context better
```

```python
import nltk

text = input("Enter a paragraph:\n")

sentences = nltk.sent_tokenize(text)

print("\nSentence Emotions:\n")

emotion_flow = []

for s in sentences:
    result = emotion_model(s)
    emotion = result[0][0]['label']
    emotion_flow.append(emotion)

    print(f"{s} → {emotion}")

print("\nEmotion Flow:")
print(" → ".join(emotion_flow))
```
